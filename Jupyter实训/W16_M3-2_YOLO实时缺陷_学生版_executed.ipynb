```python
def detect(img, win=8, thresh=0.5):
    """阈值→定位框：分数图最大点(超过阈值)映射回原图坐标"""
    smap = score_map(img, win)
    if smap is None:
        return None
    r, c = np.unravel_index(np.argmax(smap), smap.shape)
    if smap[r, c] < thresh:
        return None
    return (c * win, r * win, win, win)
```
