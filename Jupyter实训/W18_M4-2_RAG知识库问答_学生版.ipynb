### Starter 3：`def tokens(text):`
（Investigate: 读代码→Modify: 补全TODO）

```
def tokens(text):
    toks = []
    for r in re.findall(r"[\u4e00-\u9fa5]+", text):
        toks += [r[i:i + 2] for i in range(len(r) - 1)] + [r]
    return set(toks)
```
